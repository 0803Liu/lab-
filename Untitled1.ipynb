```python
import time, uuid, statistics, json
from openai import OpenAI

client = OpenAI(base_url="http://localhost:8000/v1", api_key="none")
MODEL = "Qwen/Qwen3-4B-Instruct-2507"

def timed_chat(messages, max_tokens, tag_prefix="req", **extra):
    t0 = time.perf_counter()
    stream = client.chat.completions.create(
        model=MODEL, messages=messages, max_tokens=max_tokens,
        temperature=0, stream=True,
        stream_options={"include_usage": True},
        extra_body={"ignore_eos": True, **extra},
    )
    t1 = t_last = None
    usage, text = None, []
    for chunk in stream:
        if chunk.usage is not None:
            usage = chunk.usage
        if chunk.choices and chunk.choices[0].delta.content:
            now = time.perf_counter()
            t1 = t1 or now
            t_last = now
            text.append(chunk.choices[0].delta.content)
    n = usage.completion_tokens
    return {
        "tag": tag_prefix,
        "ttft_s": t1 - t0,
        "tpot_s": (t_last - t1) / (n - 1) if n > 1 else None,
        "e2e_s": t_last - t0,
        "prompt_tokens": usage.prompt_tokens,
        "output_tokens": n,
        "text": "".join(text),
    }
```

```python
#加载 tokenizer，构造 make_prompt 与 exp1_request
from transformers import AutoTokenizer
import uuid

tok = AutoTokenizer.from_pretrained("Qwen/Qwen3-4B-Instruct-2507")

def make_prompt(target_tokens):
    ids = tok.encode(raw, add_special_tokens=False)[:target_tokens]
    return tok.decode(ids, skip_special_tokens=True)

def exp1_request(target_prompt_tokens):
    tag = f"req-{uuid.uuid4().hex[:8]}"
    body = make_prompt(target_prompt_tokens)
    content = f"{tag}\nSummarize the following text in one sentence.\n\n{body}"
    return [{"role": "user", "content": content}]

# 自检：16384 token 能切出来
test = make_prompt(16384)
print("16384 token 解码后字符数:", len(test))
```

```python
#实验 1：请求长度与 TTFT
import statistics

# 舍弃 16384，防止 Colab 断连
lengths = [256, 1024, 2048, 4096, 8192]
results_exp1 = []

print("开始跑实验 1...")
for L in lengths:
    # 热身
    timed_chat(exp1_request(L), max_tokens=16, tag_prefix=f"warmup-{L}")
    lat = []
    for i in range(5):
        r = timed_chat(exp1_request(L), max_tokens=16, tag_prefix=f"exp1-{L}-{i}")
        lat.append(r["ttft_s"])
        results_exp1.append(r)

    med = statistics.median(lat)
    print(f"prompt={L:6d}  TTFT median={med*1000:9.1f} ms  "
          f"prefill≈{L/med:8.0f} tok/s  range=[{min(lat)*1000:.0f}, {max(lat)*1000:.0f}] ms")
```

跑完后你会看到 TTFT 随长度增长、prefill tok/s 在长请求时下降

```python
#实验 1 补充：前缀缓存对比
fixed_body = make_prompt(8192)
fixed_messages = [
    {"role": "user",
     "content": f"Summarize the following text in one sentence.\n\n{fixed_body}"}
]

ttfts_prefix = []
for i in range(3):
    r = timed_chat(fixed_messages, max_tokens=16, tag_prefix=f"prefix-{i}")
    ttfts_prefix.append(r["ttft_s"])
    print(f"第 {i+1} 次: TTFT={r['ttft_s']*1000:8.1f} ms  prompt_tokens={r['prompt_tokens']}")

print(f"\n第 2 次加速: {ttfts_prefix[0]/ttfts_prefix[1]:.1f}x")
print(f"第 3 次加速: {ttfts_prefix[0]/ttfts_prefix[2]:.1f}x")
```

第 2 次 TTFT 应降到几十毫秒级——这就是 LR4 agent 24 个工具描述靠前缀缓存省下的时间

“受限于 Colab T4 的算力与显存限制，服务端的前缀缓存未能成功复用（3 次 TTFT 均约为 31 秒，无加速效果）。但从原理上，前缀缓存对于 LR4 中 24 个工具描述的固定前缀有极大的加速作用。”

```python
#实验 2：输出长度与 TPOT
import numpy as np

results_exp2 = []
print("开始跑实验 2（精简版）...")

# 只测 16 和 64 两个长度，且每个长度只跑 2 次
for out_len in [16, 64]:
    timed_chat(exp1_request(1024), max_tokens=out_len, tag_prefix="warmup-exp2")
    for i in range(2):
        print(f"  正在测输出 {out_len} token (第 {i+1}/2 次)...")
        r = timed_chat(exp1_request(1024), max_tokens=out_len,
                       tag_prefix=f"exp2-{out_len}-{i}")
        results_exp2.append(r)

# 如果只有 2 个点，我们用简单的两点求斜率，代替 polyfit
if len(results_exp2) >= 2:
    xs = [r["output_tokens"] for r in results_exp2]
    ys = [r["e2e_s"] for r in results_exp2]
    coef = np.polyfit(xs, ys, 1)
    print(f"\n拟合结果:")
    print(f"e2e = {coef[1]*1000:.0f} ms + {coef[0]*1000:.1f} ms/token")
    print(f"截距（≈TTFT）: {coef[1]*1000:.0f} ms")
    print(f"斜率（即 TPOT）: {coef[0]*1000:.1f} ms  →  {1/coef[0]:.1f} tok/s")
else:
    print("\n数据不足，但你已经有了 TTFT 和 E2E 的实测点。")
```

```python
#实验 3 前的准备：异步客户端
import asyncio
from openai import AsyncOpenAI

aclient = AsyncOpenAI(base_url="http://localhost:8000/v1", api_key="none")

async def one_client(client_id, n_requests, results_list):
    for i in range(n_requests):
        tag = f"c{client_id}-r{i}-{uuid.uuid4().hex[:6]}"
        body = make_prompt(1024)
        content = f"{tag}\nSummarize the following text in one sentence.\n\n{body}"
        messages = [{"role": "user", "content": content}]
        t0 = time.perf_counter()
        stream = await aclient.chat.completions.create(
            model=MODEL, messages=messages, max_tokens=128,
            temperature=0, stream=True,
            stream_options={"include_usage": True},
            extra_body={"ignore_eos": True},
        )
        t1 = t_last = None
        usage = None
        async for chunk in stream:
            if chunk.usage:
                usage = chunk.usage
            if chunk.choices and chunk.choices[0].delta.content:
                now = time.perf_counter()
                t1 = t1 or now
                t_last = now
        n = usage.completion_tokens
        results_list.append({
            "client": client_id,
            "ttft": t1 - t0,
            "tpot": (t_last - t1) / (n - 1) if n > 1 else None,
            "e2e": t_last - t0,
            "output_tokens": n,
        })
```

```python
#探测 /metrics 指标名
def get_metrics():
    return requests.get("http://localhost:8000/metrics").text

for line in get_metrics().splitlines():
    if "num_requests_waiting" in line or "kv_cache_usage" in line:
        print(line)
```

waiting_after > 0 或 kv_usage_after 接近 1.0 的并发级别，就是 KV 缓存饱和、开始排队的拐点，与 Cell 3 的 KV_cache_tokens // 1152 对照。

```python
#分级负载 + metrics
import re
import asyncio

def parse_metric(text, name):
    pattern = rf"^{re.escape(name)}\s+([0-9.eE+-]+)$"
    for line in text.splitlines():
        m = re.match(pattern, line.strip())
        if m:
            return float(m.group(1))
    return None

async def run_level_with_metrics(c):
    n_per_client = 2  # 每个客户端发 2 个请求
    results = []

    start = time.perf_counter()
    await asyncio.gather(*[one_client(i, n_per_client, results) for i in range(c)])
    elapsed = time.perf_counter() - start

    total_out = sum(r["output_tokens"] for r in results)
    ttfts = sorted(r["ttft"] for r in results)
    tpots = [r["tpot"] for r in results if r["tpot"]]

    return {
        "concurrency": c,
        "throughput_tok_s": total_out / elapsed,
        "ttft_median_ms": statistics.median(ttfts) * 1000,
        "tpot_median_ms": statistics.median(tpots) * 1000,
    }

async def main_with_metrics():
    all_levels = []
    print("开始跑实验 3（极简版）...")
    for c in [1, 2]:
        print(f"正在测试并发数 = {c} ...")
        r = await run_level_with_metrics(c)
        all_levels.append(r)
        print(r)
    return all_levels

all_levels = await main_with_metrics()
```

```python
#保存原始记录
def dump_jsonl(records, path):
    with open(path, "w") as f:
        for r in records:
            f.write(json.dumps(r, ensure_ascii=False) + "\n")

dump_jsonl(results_exp1, "exp1_records.jsonl")
dump_jsonl(results_exp2, "exp2_records.jsonl")
dump_jsonl(all_levels, "exp3_levels.jsonl")
print("已保存 exp1/exp2/exp3 的 JSONL")
```

```python
import matplotlib.pyplot as plt
import statistics
import numpy as np

# ==========================================
# 图 1：实验 1 (TTFT vs prompt length)
# ==========================================
by_len = {}
for r in results_exp1:
    by_len.setdefault(r["prompt_tokens"], []).append(r["ttft_s"])
lens = sorted(by_len)
med = [statistics.median(by_len[L]) * 1000 for L in lens]
lo = [min(by_len[L]) * 1000 for L in lens]
hi = [max(by_len[L]) * 1000 for L in lens]

plt.figure()
plt.plot(lens, med, "o-", label="median")
plt.fill_between(lens, lo, hi, alpha=0.2, label="range")
plt.xlabel("prompt tokens"); plt.ylabel("TTFT (ms)")
plt.title("Exp1: TTFT vs prompt length")
plt.legend(); plt.grid(True)
plt.savefig("exp1_ttft_vs_len.png", dpi=150, bbox_inches="tight")
plt.show() # 显示图 1

# ==========================================
# 图 2：实验 2 (E2E vs output length)
# ==========================================
xs2 = [r["output_tokens"] for r in results_exp2]
ys2 = [r["e2e_s"] * 1000 for r in results_exp2]
coef2 = np.polyfit(xs2, ys2, 1)
fit_x = np.linspace(min(xs2), max(xs2), 100)
fit_y = coef2[0] * fit_x + coef2[1]

plt.figure()
plt.scatter(xs2, ys2, alpha=0.6, label="measured")
plt.plot(fit_x, fit_y, "r--", label=f"fit: {coef2[1]:.0f}+{coef2[0]:.1f}N ms")
plt.xlabel("output tokens"); plt.ylabel("end-to-end (ms)")
plt.title("Exp2: e2e vs output length")
plt.legend(); plt.grid(True)
plt.savefig("exp2_e2e_vs_outlen.png", dpi=150, bbox_inches="tight")
plt.show() # 显示图 2

# ==========================================
# 图 3：实验 3 (Throughput vs Latency)
# ==========================================
cs = [r["concurrency"] for r in all_levels]
thr = [r["throughput_tok_s"] for r in all_levels]

# 【关键修复】这里把 ttft_p90_ms 换成了你实际有的 ttft_median_ms
lat = [r["ttft_median_ms"] for r in all_levels]

plt.figure()
plt.plot(thr, lat, "o-", color="purple")
for c, x, y in zip(cs, thr, lat):
    plt.annotate(f"c={c}", (x, y), textcoords="offset points", xytext=(5, 5))
plt.axhline(2000, color="red", linestyle=":", label="SLO TTFT = 2s")
plt.xlabel("throughput (output tok/s)")
plt.ylabel("TTFT median (ms)")
plt.title("Exp3: throughput vs latency")
plt.legend(); plt.grid(True)
plt.savefig("exp3_throughput_vs_latency.png", dpi=150, bbox_inches="tight")
plt.show() # 显示图 3

print("✅ 三张图已全部生成并保存！请查看 Colab 左侧文件栏。")
```
